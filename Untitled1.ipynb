```python
# ============================================
# EXPLORATORY DATA ANALYSIS (EDA) PROJECT
# Student Performance Dataset
# ============================================

# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")


# ============================================
# Step 2: Create Dataset
# ============================================

data = {
    "Student_ID": range(1, 21),
    "Age": [18,19,18,20,19,18,21,20,19,18,
            20,21,19,18,20,19,21,18,20,19],
    "Study_Hours": [2,4,3,5,6,2,7,5,4,3,
                    6,8,5,2,7,4,9,3,6,5],
    "Attendance": [70,80,75,90,85,65,95,88,82,72,
                   91,96,86,68,94,81,98,74,89,87],
    "Assignments": [5,7,6,9,8,4,10,9,7,6,
                    9,10,8,5,10,7,10,6,9,8],
    "Exam_Score": [55,68,61,82,76,50,90,84,72,60,
                   85,94,79,52,91,70,96,63,88,81]
}

df = pd.DataFrame(data)

print("\nDataset:")
print(df)


# ============================================
# Step 3: Display First and Last Records
# ============================================

print("\nFirst 5 Records:")
print(df.head())

print("\nLast 5 Records:")
print(df.tail())


# ============================================
# Step 4: Dataset Information
# ============================================

print("\nDataset Information:")
df.info()


# ============================================
# Step 5: Dataset Shape
# ============================================

print("\nNumber of Rows and Columns:")
print(df.shape)


# ============================================
# Step 6: Check Missing Values
# ============================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================
# Step 7: Check Duplicate Values
# ============================================

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================
# Step 8: Statistical Summary
# ============================================

print("\nStatistical Summary:")
print(df.describe())


# ============================================
# Step 9: Mean, Median and Standard Deviation
# ============================================

print("\nMean:")
print(df.mean(numeric_only=True))

print("\nMedian:")
print(df.median(numeric_only=True))

print("\nStandard Deviation:")
print(df.std(numeric_only=True))


# ============================================
# Step 10: Correlation Analysis
# ============================================

correlation = df.corr(numeric_only=True)

print("\nCorrelation Matrix:")
print(correlation)


# ============================================
# Step 11: Correlation Heatmap
# ============================================

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()


# ============================================
# Step 12: Study Hours vs Exam Score
# ============================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.show()


# ============================================
# Step 13: Attendance vs Exam Score
# ============================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam_Score"
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")

plt.show()


# ============================================
# Step 14: Exam Score Distribution
# ============================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Exam_Score"],
    bins=6,
    kde=True
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()


# ============================================
# Step 15: Study Hours Distribution
# ============================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Study_Hours"],
    bins=8,
    kde=True
)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()


# ============================================
# Step 16: Boxplot for Exam Scores
# ============================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    y=df["Exam_Score"]
)

plt.title("Exam Score Boxplot")
plt.ylabel("Exam Score")

plt.show()


# ============================================
# Step 17: Average Exam Score
# ============================================

average_score = df["Exam_Score"].mean()

print("\nAverage Exam Score:", round(average_score, 2))


# ============================================
# Step 18: Highest and Lowest Score
# ============================================

highest_score = df["Exam_Score"].max()
lowest_score = df["Exam_Score"].min()

print("Highest Exam Score:", highest_score)
print("Lowest Exam Score:", lowest_score)


# ============================================
# Step 19: Top Performing Students
# ============================================

top_students = df.sort_values(
    by="Exam_Score",
    ascending=False
).head(5)

print("\nTop 5 Performing Students:")
print(top_students)


# ============================================
# Step 20: Final Insights
# ============================================

print("\n========== EDA INSIGHTS ==========")

print("1. Students with more study hours generally have higher exam scores.")

print("2. Higher attendance is generally associated with better exam performance.")

print("3. Assignment completion also shows a positive relationship with exam scores.")

print("4. Correlation analysis helps identify the factors most related to exam performance.")

print("5. Statistical summaries and visualizations help understand the overall dataset.")

print("\n========== EDA PROJECT COMPLETED ==========")
```
